# Veri Filtreleme & Sıralama Projesi
## Dünya Ülkeleri Veri Analizi

Bu çalışmada ülke verilerini filtreleyip sıralayarak anlamlı bilgiler çıkaracağız.

**Pandas'ın Temel Fonksiyonları:**
- `sort_values()`: Sıralama
- `loc[]` / `iloc[]`: Filtreleme
- `nlargest()` / `nsmallest()`: En büyük/küçük n değer

**Görevler:**
1. Nüfusa göre azalan sırada ülkeleri listele
2. GDP per capita'ya göre artan sırada ülkeleri listele
3. Nüfusu 10 milyonun üzerinde olan ülkeleri seç
4. Literacy oranına göre en yüksek ilk 5 ülkeyi bul
5. GDP per capita > 10.000 olan ülkeleri filtrele
6. Nüfus yoğunluğu en yüksek ilk 10 ülkeyi bul

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Pandas görüntüleme ayarları
pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.float_format', '{:.2f}'.format)

%matplotlib inline
print("✓ Kütüphaneler yüklendi!")
```

## Başlangıç: Veriyi Yükleme ve İnceleme

Öncelikle veri setimizi tanıyalım.

```python
print("📂 VERİ YÜKLEME")
print("="*80)

# CSV dosyasını yükle
df = pd.read_csv('country.csv')

print(f"✓ Veri seti yüklendi!")
print(f"   Boyut: {df.shape[0]} ülke × {df.shape[1]} özellik\n")

print("="*80)
print("📊 İLK 10 ÜLKE:")
print("="*80)
print(df.head(10))

print("\n" + "="*80)
print("📋 KOLON BİLGİLERİ:")
print("="*80)
print(df.info())

print("\n" + "="*80)
print("📈 İSTATİSTİKSEL ÖZET:")
print("="*80)
print(df.describe())

print("\n" + "="*80)
print("📝 KOLON İSİMLERİ:")
print("="*80)
for i, col in enumerate(df.columns, 1):
    print(f"   {i}. {col}")

# Eksik değer kontrolü
print("\n" + "="*80)
print("🔍 EKSİK DEĞER KONTROLÜ:")
print("="*80)
eksik = df.isnull().sum()
if eksik.sum() > 0:
    print(eksik[eksik > 0])
else:
    print("✓ Eksik değer bulunamadı!")
```

## Görev 1: Nüfusa Göre Azalan Sırada Ülkeleri Listele

**Amaç:** En kalabalık ülkeleri bulmak

**Kullanılan Fonksiyon:** `sort_values(ascending=False)`

```python
print("\n" + "="*80)
print("📊 GÖREV 1: NÜFUSA GÖRE AZALAN SIRADA ÜLKELER")
print("="*80)

# Nüfusa göre azalan sırada sırala
nufusa_gore = df.sort_values('Population', ascending=False)

print("\n🏆 EN KALABALIK 15 ÜLKE:")
print("="*80)
top15_nufus = nufusa_gore[['Country', 'Population', 'Region']].head(15)
top15_nufus.index = range(1, 16)  # İndeksi 1'den başlat
print(top15_nufus)

print(f"\n📈 İSTATİSTİKLER:")
print(f"   En kalabalık: {nufusa_gore.iloc[0]['Country']} ({nufusa_gore.iloc[0]['Population']:,.0f})")
print(f"   En az kalabalık: {nufusa_gore.iloc[-1]['Country']} ({nufusa_gore.iloc[-1]['Population']:,.0f})")
print(f"   Ortalama nüfus: {df['Population'].mean():,.0f}")

# Görselleştirme
plt.figure(figsize=(14, 8))
top10_countries = nufusa_gore.head(10)
plt.barh(top10_countries['Country'], top10_countries['Population'],
         color='steelblue', edgecolor='black', linewidth=1.5)
plt.xlabel('Nüfus', fontsize=12, fontweight='bold')
plt.ylabel('Ülke', fontsize=12, fontweight='bold')
plt.title('En Kalabalık 10 Ülke', fontsize=15, fontweight='bold', pad=20)
plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```

## Görev 2: GDP per Capita'ya Göre Artan Sırada Ülkeleri Listele

**Amaç:** En düşük kişi başı gelirden en yükseğe doğru sıralama


```python
print("\n" + "="*80)
print("📊 GÖREV 2: GDP PER CAPITA'YA GÖRE ARTAN SIRADA ÜLKELER")
print("="*80)

# GDP per capita'ya göre artan sırada sırala
gdp_artan = df.sort_values('GDP ($ per capita)', ascending=True)

print("\n💰 EN DÜŞÜK GELİRLİ 10 ÜLKE:")
print("="*80)
bottom10_gdp = gdp_artan[['Country', 'GDP ($ per capita)', 'Region']].head(10)
bottom10_gdp.index = range(1, 11)
print(bottom10_gdp)

print("\n💎 EN YÜKSEK GELİRLİ 10 ÜLKE:")
print("="*80)
top10_gdp = gdp_artan[['Country', 'GDP ($ per capita)', 'Region']].tail(10)
top10_gdp = top10_gdp.sort_values('GDP ($ per capita)', ascending=False)
top10_gdp.index = range(1, 11)
print(top10_gdp)

print(f"\n📊 İSTATİSTİKLER:")
print(f"   En düşük GDP: {gdp_artan.iloc[0]['Country']} (${gdp_artan.iloc[0]['GDP ($ per capita)']:,.0f})")
print(f"   En yüksek GDP: {gdp_artan.iloc[-1]['Country']} (${gdp_artan.iloc[-1]['GDP ($ per capita)']:,.0f})")
print(f"   Ortalama GDP: ${df['GDP ($ per capita)'].mean():,.0f}")

# Görselleştirme - En zengin ülkeler
plt.figure(figsize=(14, 8))
top10_rich = df.nlargest(10, 'GDP ($ per capita)')
plt.barh(top10_rich['Country'], top10_rich['GDP ($ per capita)'],
         color='gold', edgecolor='black', linewidth=1.5)
plt.xlabel('GDP per Capita ($)', fontsize=12, fontweight='bold')
plt.ylabel('Ülke', fontsize=12, fontweight='bold')
plt.title('En Yüksek Kişi Başı Gelire Sahip 10 Ülke', fontsize=15, fontweight='bold', pad=20)
plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```

## Görev 3: Nüfusu 10 Milyonun Üzerinde Olan Ülkeleri Seç

**Amaç:** Büyük nüfuslu ülkeleri filtrelemek

**Kullanılan Yöntem:** Boolean indexing (df[koşul])

```python
print("\n" + "="*80)
print("📊 GÖREV 3: NÜFUSU 10 MİLYONUN ÜZERİNDE OLAN ÜLKELER")
print("="*80)

# Nüfusu 10 milyonun üzerinde olan ülkeleri filtrele
nufus_10m = df[df['Population'] > 10_000_000].copy()
nufus_10m = nufus_10m.sort_values('Population', ascending=False)

print(f"\n✓ Toplam {len(nufus_10m)} ülke bulundu!")
print(f"   (Tüm ülkelerin %{len(nufus_10m)/len(df)*100:.1f}'i)")

print("\n📋 NÜFUSU 10 MİLYON ÜSTÜ ÜLKELER (İlk 20):")
print("="*80)
display_cols = ['Country', 'Population', 'Region', 'GDP ($ per capita)']
result = nufus_10m[display_cols].head(20)
result.index = range(1, len(result)+1)
print(result)

# Bölgesel dağılım
print("\n🌍 BÖLGESEL DAĞILIM:")
print("="*80)
bolgesel = nufus_10m['Region'].value_counts()
print(bolgesel)

# Görselleştirme
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Sol: Bölgesel dağılım
bolgesel.plot(kind='bar', ax=axes[0], color='coral', edgecolor='black', linewidth=1.5)
axes[0].set_title('Büyük Nüfuslu Ülkelerin Bölgesel Dağılımı', fontweight='bold', fontsize=13)
axes[0].set_xlabel('Bölge', fontweight='bold')
axes[0].set_ylabel('Ülke Sayısı', fontweight='bold')
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(axis='y', alpha=0.3)

# Sağ: Top 10 nüfus
top10 = nufus_10m.head(10)
axes[1].barh(top10['Country'], top10['Population'], color='skyblue', 
             edgecolor='black', linewidth=1.5)
axes[1].set_title('En Kalabalık 10 Ülke (>10M)', fontweight='bold', fontsize=13)
axes[1].set_xlabel('Nüfus', fontweight='bold')
axes[1].invert_yaxis()
axes[1].grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Görev 4: Literacy Oranına Göre En Yüksek İlk 5 Ülkeyi Seç

**Amaç:** En yüksek okuma-yazma oranına sahip ülkeleri bulmak

**Kullanılan Fonksiyon:** `nlargest(n, column)`

```python
print("\n" + "="*80)
print("📊 GÖREV 4: LITERACY ORANINA GÖRE EN YÜKSEK İLK 5 ÜLKE")
print("="*80)

# En yüksek literacy oranına sahip 5 ülke
literacy_top5 = df.nlargest(5, 'Literacy (%)')

print("\n🎓 EN YÜKSEK OKUMA-YAZMA ORANINA SAHİP 5 ÜLKE:")
print("="*80)
display_cols = ['Country', 'Literacy (%)', 'Region', 'GDP ($ per capita)']
result = literacy_top5[display_cols].copy()
result.index = range(1, 6)
print(result)

# Ek analiz - Top 15
print("\n📚 EN YÜKSEK OKUMA-YAZMA ORANINA SAHİP 15 ÜLKE:")
print("="*80)
literacy_top15 = df.nlargest(15, 'Literacy (%)')
result15 = literacy_top15[display_cols].copy()
result15.index = range(1, 16)
print(result15)

print(f"\n📊 İSTATİSTİKLER:")
print(f"   En yüksek: {literacy_top5.iloc[0]['Country']} ({literacy_top5.iloc[0]['Literacy (%)']:.1f}%)")
print(f"   Dünya ortalaması: {df['Literacy (%)'].mean():.2f}%")
print(f"   Medyan: {df['Literacy (%)'].median():.2f}%")

# Görselleştirme
plt.figure(figsize=(14, 8))
top20_literacy = df.nlargest(20, 'Literacy (%)')
colors = plt.cm.Greens(np.linspace(0.4, 0.9, 20))
plt.barh(top20_literacy['Country'], top20_literacy['Literacy (%)'],
         color=colors, edgecolor='black', linewidth=1.5)
plt.xlabel('Okuma-Yazma Oranı (%)', fontsize=12, fontweight='bold')
plt.ylabel('Ülke', fontsize=12, fontweight='bold')
plt.title('En Yüksek Okuma-Yazma Oranına Sahip 20 Ülke', fontsize=15, fontweight='bold', pad=20)
plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3, linestyle='--')
plt.axvline(x=df['Literacy (%)'].mean(), color='red', linestyle='--', 
           linewidth=2, label=f"Dünya Ort. ({df['Literacy (%)'].mean():.1f}%)")
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

```

## Görev 5: GDP per Capita > 10.000 Olan Ülkeleri Filtrele

**Amaç:** Gelişmiş ekonomilere sahip ülkeleri seçmek

**Kritik Eşik:** $10,000 - Genelde gelişmiş/gelişmekte olan ayırımı

```python
print("\n" + "="*80)
print("📊 GÖREV 5: GDP PER CAPITA > $10,000 OLAN ÜLKELER")
print("="*80)

# GDP per capita > 10000 olan ülkeleri filtrele
gdp_10k = df[df['GDP ($ per capita)'] > 10_000].copy()
gdp_10k = gdp_10k.sort_values('GDP ($ per capita)', ascending=False)

print(f"\n✓ Toplam {len(gdp_10k)} ülke bulundu!")
print(f"   (Tüm ülkelerin %{len(gdp_10k)/len(df)*100:.1f}'i)")

print("\n💰 GDP > $10,000 OLAN ÜLKELER (İlk 20):")
print("="*80)
display_cols = ['Country', 'GDP ($ per capita)', 'Region', 'Population']
result = gdp_10k[display_cols].head(20)
result.index = range(1, len(result)+1)
print(result)

# Bölgesel analiz
print("\n🌍 BÖLGESEL DAĞILIM:")
print("="*80)
bolgesel_gdp = gdp_10k['Region'].value_counts()
print(bolgesel_gdp)

# İstatistikler
print(f"\n📊 İSTATİSTİKLER:")
print(f"   En yüksek GDP: {gdp_10k.iloc[0]['Country']} (${gdp_10k.iloc[0]['GDP ($ per capita)']:,.0f})")
print(f"   Bu grubun ortalama GDP'si: ${gdp_10k['GDP ($ per capita)'].mean():,.0f}")
print(f"   Bu grubun toplam nüfusu: {gdp_10k['Population'].sum():,.0f}")

# Görselleştirme
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Bölgesel dağılım
bolgesel_gdp.plot(kind='bar', ax=axes[0, 0], color='purple', 
                  edgecolor='black', linewidth=1.5)
axes[0, 0].set_title('Zengin Ülkelerin Bölgesel Dağılımı (GDP > $10k)', 
                     fontweight='bold', fontsize=12)
axes[0, 0].set_xlabel('Bölge', fontweight='bold')
axes[0, 0].set_ylabel('Ülke Sayısı', fontweight='bold')
axes[0, 0].tick_params(axis='x', rotation=45)
axes[0, 0].grid(axis='y', alpha=0.3)

# 2. Top 15 GDP
top15_gdp = gdp_10k.head(15)
axes[0, 1].barh(top15_gdp['Country'], top15_gdp['GDP ($ per capita)'],
                color='gold', edgecolor='black', linewidth=1.5)
axes[0, 1].set_title('En Zengin 15 Ülke', fontweight='bold', fontsize=12)
axes[0, 1].set_xlabel('GDP per Capita ($)', fontweight='bold')
axes[0, 1].invert_yaxis()
axes[0, 1].grid(axis='x', alpha=0.3)

# 3. GDP vs Nüfus scatter
axes[1, 0].scatter(gdp_10k['Population'], gdp_10k['GDP ($ per capita)'],
                   alpha=0.6, s=100, c='green', edgecolors='black', linewidth=0.5)
axes[1, 0].set_title('GDP vs Nüfus İlişkisi', fontweight='bold', fontsize=12)
axes[1, 0].set_xlabel('Nüfus', fontweight='bold')
axes[1, 0].set_ylabel('GDP per Capita ($)', fontweight='bold')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].set_xscale('log')

# 4. Pasta grafiği - Bölgesel
axes[1, 1].pie(bolgesel_gdp.values, labels=bolgesel_gdp.index,
               autopct='%1.1f%%', startangle=90, 
               colors=plt.cm.Set3(np.linspace(0, 1, len(bolgesel_gdp))))
axes[1, 1].set_title('Zengin Ülkelerin Bölgesel Oranı', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.show()
```

## Görev 6: Nüfus Yoğunluğu En Yüksek İlk 10 Ülkeyi Bul

**Amaç:** En kalabalık yaşanan ülkeleri bulmak

**Nüfus Yoğunluğu:** Nüfus / Alan (kişi/km²)

**Not:** Bazı ülkelerde alan bilgisi eksik olabilir!
"""

```python
print("\n" + "="*80)
print("📊 GÖREV 6: NÜFUS YOĞUNLUĞU EN YÜKSEK İLK 10 ÜLKE")
print("="*80)

# Nüfus yoğunluğunu hesapla
df['Nufus_Yogunlugu'] = df['Population'] / df['Area (sq. mi.)']

# Sonsuz ve NaN değerleri temizle
df_yogunluk = df[np.isfinite(df['Nufus_Yogunlugu'])].copy()

print(f"\n✓ {len(df_yogunluk)} ülke için yoğunluk hesaplandı")
print(f"   (Alan bilgisi eksik olan {len(df) - len(df_yogunluk)} ülke hariç tutuldu)")

# En yüksek nüfus yoğunluğuna sahip 10 ülke
yogunluk_top10 = df_yogunluk.nlargest(10, 'Nufus_Yogunlugu')

print("\n🏙️ EN YÜKSEK NÜFUS YOĞUNLUĞUNA SAHİP 10 ÜLKE:")
print("="*80)
display_cols = ['Country', 'Population', 'Area (sq. mi.)', 'Nufus_Yogunlugu', 'Region']
result = yogunluk_top10[display_cols].copy()
result.index = range(1, 11)
print(result)

# Top 20 liste
print("\n🏘️ EN YÜKSEK NÜFUS YOĞUNLUĞUNA SAHİP 20 ÜLKE:")
print("="*80)
yogunluk_top20 = df_yogunluk.nlargest(20, 'Nufus_Yogunlugu')
result20 = yogunluk_top20[display_cols].copy()
result20.index = range(1, 21)
print(result20)

print(f"\n📊 İSTATİSTİKLER:")
print(f"   En yüksek: {yogunluk_top10.iloc[0]['Country']} ({yogunluk_top10.iloc[0]['Nufus_Yogunlugu']:.0f} kişi/km²)")
print(f"   Dünya ortalaması: {df_yogunluk['Nufus_Yogunlugu'].mean():.2f} kişi/km²")
print(f"   Medyan: {df_yogunluk['Nufus_Yogunlugu'].median():.2f} kişi/km²")

# Görselleştirme
fig, axes = plt.subplots(2, 1, figsize=(14, 12))

# 1. Top 15 nüfus yoğunluğu
top15_yog = df_yogunluk.nlargest(15, 'Nufus_Yogunlugu')
colors = plt.cm.Reds(np.linspace(0.4, 0.9, 15))
axes[0].barh(top15_yog['Country'], top15_yog['Nufus_Yogunlugu'],
             color=colors, edgecolor='black', linewidth=1.5)
axes[0].set_xlabel('Nüfus Yoğunluğu (kişi/km²)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Ülke', fontsize=12, fontweight='bold')
axes[0].set_title('En Yüksek Nüfus Yoğunluğuna Sahip 15 Ülke', 
                 fontsize=14, fontweight='bold', pad=15)
axes[0].invert_yaxis()
axes[0].grid(axis='x', alpha=0.3, linestyle='--')

# 2. Nüfus yoğunluğu dağılımı (histogram)
axes[1].hist(df_yogunluk['Nufus_Yogunlugu'], bins=50, color='teal', 
            edgecolor='black', linewidth=1, alpha=0.7)
axes[1].set_xlabel('Nüfus Yoğunluğu (kişi/km²)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Ülke Sayısı', fontsize=12, fontweight='bold')
axes[1].set_title('Dünya Geneli Nüfus Yoğunluğu Dağılımı', 
                 fontsize=14, fontweight='bold', pad=15)
axes[1].axvline(df_yogunluk['Nufus_Yogunlugu'].mean(), color='red', 
               linestyle='--', linewidth=2, 
               label=f"Ortalama ({df_yogunluk['Nufus_Yogunlugu'].mean():.0f})")
axes[1].axvline(df_yogunluk['Nufus_Yogunlugu'].median(), color='orange', 
               linestyle='--', linewidth=2, 
               label=f"Medyan ({df_yogunluk['Nufus_Yogunlugu'].median():.0f})")
axes[1].legend(fontsize=11)
axes[1].grid(axis='y', alpha=0.3)
axes[1].set_xlim(0, 1000)  # Görülebilirlik için limit

plt.tight_layout()
plt.show()
```
