```python
# ÖDEV 1: VERİ GÖRSELLEŞTİRME (50_Startups.csv)

"""
# Veri Görselleştirme Ödevi
## 50_Startups.csv Veri Seti Analizi

Bu çalışmada startup şirketlerinin harcama ve kâr verilerini analiz edeceğiz.

**Görevler:**
1. R&D harcaması ile kâr ilişkisi
2. Yönetim harcaması ile kâr ilişkisi  
3. Eyaletlere göre ortalama kâr karşılaştırması
4. Harcama türlerinin dağılımı
"""
```

```python

# Veri işleme için pandas
import pandas as pd

# Görselleştirme için matplotlib ve seaborn
import matplotlib.pyplot as plt
import seaborn as sns

# Grafiklerin notebook içinde görünmesi için
%matplotlib inline

# Grafik stilini ayarlayalım
sns.set_style("whitegrid")
plt.rcParams['figure.dpi'] = 100

print("✓ Kütüphaneler başarıyla yüklendi!")
```

```python

#Öncelikle veri setimizi yükleyip yapısını inceleyelim.
# CSV dosyasını okuyalım
df = pd.read_csv('50_Startups.csv')

# İlk 5 satırı görüntüleyelim
print("📊 İlk 5 Satır:")
print(df.head())

print("\n" + "="*70 + "\n")

# Veri seti hakkında genel bilgi
print("📋 Veri Seti Bilgileri:")
print(df.info())

print("\n" + "="*70 + "\n")

# İstatistiksel özet
print("📈 İstatistiksel Özet:")
print(df.describe())

print("\n" + "="*70 + "\n")

# Kolon isimleri
print("📝 Kolonlar:", df.columns.tolist())
```

```python
"""
---
## Görev 1: R&D Harcaması ile Kâr Arasındaki İlişki

R&D (Araştırma ve Geliştirme) harcamalarının şirket kârı üzerindeki etkisini 
scatter plot (dağılım grafiği) ile görselleştireceğiz.

**Hipotez:** R&D harcaması arttıkça kâr da artmalı.
"""
# Grafik boyutunu ayarla
plt.figure(figsize=(12, 7))

# Scatter plot oluştur
plt.scatter(df['R&D Spend'], df['Profit'], 
           alpha=0.6,           # Saydamlık
           color='dodgerblue',  # Renk
           s=100,               # Nokta boyutu
           edgecolors='black',  # Kenar rengi
           linewidth=0.5)       # Kenar kalınlığı

# Eksen etiketleri
plt.xlabel('R&D Harcaması ($)', fontsize=13, fontweight='bold')
plt.ylabel('Kâr ($)', fontsize=13, fontweight='bold')

# Başlık
plt.title('R&D Harcaması ile Kâr Arasındaki İlişki', 
         fontsize=15, fontweight='bold', pad=20)

# Grid çizgileri
plt.grid(True, alpha=0.3, linestyle='--')

# Düzeni ayarla ve göster
plt.tight_layout()
plt.show()

# Korelasyon hesapla
korelasyon = df['R&D Spend'].corr(df['Profit'])
print(f"\n📊 Korelasyon Katsayısı: {korelasyon:.4f}")
print(f"{'Yorum: Pozitif güçlü ilişki var!' if korelasyon > 0.7 else 'Yorum: Zayıf ilişki'}")
```

```python
"""
---
## Görev 2: Yönetim Harcaması ile Kâr Arasındaki İlişki

Yönetim (Administration) harcamalarının kâr üzerindeki etkisini inceleyelim.

**Beklenti:** Yönetim harcamaları R&D kadar etkili olmayabilir.
"""
plt.figure(figsize=(12, 7))

# Scatter plot
plt.scatter(df['Administration'], df['Profit'], 
           alpha=0.6, 
           color='mediumseagreen',
           s=100,
           edgecolors='black',
           linewidth=0.5)

plt.xlabel('Yönetim Harcaması ($)', fontsize=13, fontweight='bold')
plt.ylabel('Kâr ($)', fontsize=13, fontweight='bold')
plt.title('Yönetim Harcaması ile Kâr Arasındaki İlişki', 
         fontsize=15, fontweight='bold', pad=20)
plt.grid(True, alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()

# Korelasyon
korelasyon2 = df['Administration'].corr(df['Profit'])
print(f"\n📊 Korelasyon Katsayısı: {korelasyon2:.4f}")
print("💡 Yorum: Yönetim harcaması ile kâr arasında zayıf bir ilişki var.")
```

```python
"""
---
## Görev 3: Eyaletlere Göre Ortalama Kâr Karşılaştırması

Her eyaletteki şirketlerin ortalama kârlarını bar chart ile karşılaştıralım.

**Soru:** Hangi eyalet daha karlı?
"""
 Eyaletlere göre ortalama kârı hesapla
ortalama_kar = df.groupby('State')['Profit'].mean().sort_values(ascending=False)

print("📊 Eyaletlere Göre Ortalama Kârlar:")
print(ortalama_kar)
print("\n")

# Bar chart
plt.figure(figsize=(12, 7))
bars = plt.bar(ortalama_kar.index, ortalama_kar.values, 
               color=['#FF6B6B', '#4ECDC4', '#45B7D1'],
               edgecolor='black',
               linewidth=1.5)

# En yüksek barı vurgula
bars[0].set_color('#FFD93D')
bars[0].set_edgecolor('black')
bars[0].set_linewidth(2)

plt.xlabel('Eyalet', fontsize=13, fontweight='bold')
plt.ylabel('Ortalama Kâr ($)', fontsize=13, fontweight='bold')
plt.title('Eyaletlere Göre Ortalama Kâr Karşılaştırması', 
         fontsize=15, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3, linestyle='--')

# Değer etiketleri ekle
for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
            f'${height:,.0f}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

print(f"🏆 En karlı eyalet: {ortalama_kar.index[0]} (${ortalama_kar.values[0]:,.2f})")
```

```python
"""
---
## Görev 4: Harcama Türlerinin Dağılımı (Boxplot)

R&D, Yönetim ve Pazarlama harcamalarının dağılımını kutu grafiği ile karşılaştıralım.

**Amaç:** Hangi harcama türü daha değişken? Outlier var mı?
"""
# İlgili kolonları seç
harcamalar = df[['R&D Spend', 'Administration', 'Marketing Spend']]

# Kolon isimlerini Türkçeleştir (görselleştirme için)
harcamalar.columns = ['R&D', 'Yönetim', 'Pazarlama']

# Boxplot oluştur
plt.figure(figsize=(14, 8))
box = plt.boxplot([harcamalar['R&D'], 
                    harcamalar['Yönetim'], 
                    harcamalar['Pazarlama']],
                   labels=['R&D', 'Yönetim', 'Pazarlama'],
                   patch_artist=True,
                   notch=True,
                   showmeans=True)

# Renklendirme
colors = ['#FF6B6B', '#4ECDC4', '#95E1D3']
for patch, color in zip(box['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

plt.ylabel('Harcama Miktarı ($)', fontsize=13, fontweight='bold')
plt.title('R&D, Yönetim ve Pazarlama Harcamalarının Dağılımı', 
         fontsize=15, fontweight='bold', pad=20)
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()

# İstatistiksel özet
print("\n📊 Harcamaların İstatistiksel Özeti:")
print(harcamalar.describe())

print("\n💡 YORUM:")
print("- R&D harcamaları en geniş aralığa sahip (en değişken)")
print("- Yönetim harcamaları en dar aralıkta (en tutarlı)")
print("- Kutu grafikteki noktalar aykırı değerleri (outliers) gösteriyor")
```
